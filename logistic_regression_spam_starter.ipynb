## Download the Dataset

This cell downloads `README.txt`, `data.txt`, and `labels.txt` into a local `data/spam-email/` folder.

````python
DATA_DIR = Path("data/spam-email")
DATA_DIR.mkdir(parents=True, exist_ok=True)

BASE_URL = "https://raw.githubusercontent.com/judyzhu-gmu/CYSE499_650_Summer2026/main/data/spam-email"
FILES = ["README.txt", "data.txt", "labels.txt"]

for filename in FILES:
    destination = DATA_DIR / filename
    if not destination.exists():
        urlretrieve(f"{BASE_URL}/{filename}", destination)
    print(f"Ready: {destination}")
````
